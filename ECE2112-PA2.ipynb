## A. REPRODUCIBLE NORMALIZATION PROBLEM

Create a reproducible 5x5 random integer array using `ndarray`, then normalize it using the z-score formula so that the resulting array has a mean of 0 and a standard deviation of 1.

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5)) 

# Compute the mean and standard deviation of all 25 elements
X_mean = np.mean(X)
X_std = np.std(X)

# Use the z-score formula to normalize the array
X_normalized = (X - X_mean) / X_std

# Display the original array
print("X:\n", X)

# Display the normalized array
print("\nX_normalized:\n", X_normalized)

# Display the mean and standard deviation of the normalized array as a check
# These should be approximately 0 and 1, respectively
print("\nMean of X_normalized:", X_normalized.mean())
print("\nStd of X_normalized:", X_normalized.std())
```

```python
np.save("X_normalized.npy", X_normalized)
```

## B. CUBES DIVISIBLE BY 4 PROBLEM

Build a 10x10 array of the cubes of the first 100 positive integers, then use Boolean indexing to select only the cubes that are divisible by 4.

```python
# Create the first 100 positive integers, then cube it
C = np.arange(1, 101) ** 3

# Reshape the 1D array of 100 cubes into a 10x10 array
C = C.reshape(10, 10)

# True wherever an element is divisible by 4
div_by_4 = C[C % 4 == 0]

# Display the shape of C 
print("Shape of C:", C.shape)

# Display the selected values
print("\nArray divided by 4:\n", div_by_4)

# Display the number of selected elements
print("\nNumber of selected elements:", div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```
